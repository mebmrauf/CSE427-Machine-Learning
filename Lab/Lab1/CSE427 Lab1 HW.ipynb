### Numpy

```python
import numpy as np
```

```python
a = np.array(np.random.randint(1, 101, size=20))
b = np.array(np.random.randint(1, 101, size=20))
print(a, '\n', b)
```

```python
a = a.reshape(5, 4)
b = b.reshape(5, 4)
print(a, "\n", b)
```

```python
a = a * 5
b = b * 7
print(a, "\n", b)
```

```python
b = b.T
c = a @ b
print(c)
```

```python
print(c[:, 1:3])
print(c[:])
```

```python
print(np.max(c))
findex = int(np.argmax(c))
rows, cols = c.shape
row = findex // cols
col = findex % cols
print((row, col))
```

```python
farray = c.flatten()
print(farray)
```

### Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]
plt.scatter(temperature_dhaka, humidity_dhaka, color = "red", marker = "*")
plt.title("Temperature vs Humidity in Dhaka")
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.grid()
plt.show()
```

```python
study_hours = [2,3,4,4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
plt.figure(figsize=(12,8))
plt.plot(study_hours, marks, color="black", marker='o')
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()
plt.show()
```

```python
subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
marks = [89, 90, 45, 78, 99]
plt.barh(subjects, marks, color='brown')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.title('Subjects vs Marks')
plt.show()
```

```python
colors = ['black', 'skyblue', 'gray', 'orange', 'brown']
plt.bar(subjects, marks, color=colors)
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Subjects vs Marks')
plt.show()
```

```python
stumarks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]
submarks = [89, 90, 45, 78, 99]

plt.figure(figsize=(15,10))
plt.subplot(2,2,1)
plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.title('Temperature vs Humidity')
plt.xlabel('Temperature')
plt.ylabel('Humidity')
plt.grid(True)

plt.subplot(2,2,2)
plt.plot(study_hours, stumarks, color = "black" , marker='o')
plt.title("Study Hours vs Marks")
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid()

plt.subplot(2,2,3)
plt.barh(subjects, submarks, color="brown")
plt.title('Subjects vs Marks')
plt.xlabel('Marks')
plt.ylabel('Subjects')

plt.subplot(2,2,4)
plt.bar(subjects, submarks, color=colors)
plt.title('Subjects vs Marks')
plt.xlabel('Marks')
plt.ylabel('Subjects')
plt.show()
```
